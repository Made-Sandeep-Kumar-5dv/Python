Fuel in Car
A car has 45 liters of fuel.
It consumes 18 liters during a trip.
Then the driver refuels 22 liters.

```python
# Update the fuel level after each step.
fuel_in_liters = 45
trip = 18
consumed_fuel = fuel_in_liters - trip
print(consumed_fuel) #27
refuel = 22
total_fuel_in_liters = consumed_fuel + refuel
print(total_fuel_in_liters) #49
```
