Pizza Party
There are 8 pizzas, each having 6 slices. If 9 friends share them equally,

```python
# How many slices does each person get?
pizzas = 8
slices = 6
friends = 9
total_slices_per_person = (pizzas*slices) // friends
print(total_slices_per_person) # 5

# How many slices remain?
total_slices_per_person = (pizzas*slices) % friends
print(total_slices_per_person) # 3
```

Classroom Arrangement 
There are 137 students. Each classroom can accommodate 30 students.

```python
# How many classrooms are completely filled?
students = 137
students_per_class = 30
total_classrooms = students // students_per_class
print(total_classrooms) # 4
# How many students are left for another classroom?
students_for_new_classroom = students % students_per_class
print(students_for_new_classroom) # 17
```

Water Tank

```python
# After 7 days, how much water remains?
water_in_liters = 2500
water_used_in_liters = 185
water_after_7days = water_used_in_liters * 7
print(water_after_7days) # 1295
remaining_water = water_in_liters - water_after_7days
print(remaining_water) # 1205 
```
